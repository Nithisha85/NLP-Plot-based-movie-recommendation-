```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
```

```python
# Load movie dataset
movies = pd.read_csv("../datasets/tmdb_5000_movies.csv")

# Load book dataset
books = pd.read_csv("../datasets/books_enriched.csv")

print("Datasets loaded successfully!")
```

```python
print("MOVIE DATASET")
print("Shape:", movies.shape)

print("\nBOOK DATASET")
print("Shape:", books.shape)
```

```python
print("MOVIE COLUMNS:")
print(movies.columns.tolist())

print("\nBOOK COLUMNS:")
print(books.columns.tolist())
```

```python
print("MOVIE DATASET:")
display(movies.head())

print("\nBOOK DATASET:")
display(books.head())
```

```python
print("MOVIE MISSING VALUES")
print(movies.isnull().sum())

print("\nBOOK MISSING VALUES")
print(books.isnull().sum())
```

```python
print("MOVIE MISSING VALUE PERCENTAGE")
print((movies.isnull().sum() / len(movies)) * 100)

print("\nBOOK MISSING VALUE PERCENTAGE")
print((books.isnull().sum() / len(books)) * 100)
```

```python
print("Movie duplicates:", movies.duplicated().sum())

print("Book duplicates:", books.duplicated().sum())
```

```python
print(movies['genres'].head(10))
```

```python
# Important movie columns
movie_columns = [
    'id', 'title', 'overview', 'genres',
    'release_date', 'vote_average', 'popularity'
]

# Important book columns
book_columns = [
    'book_id', 'title', 'authors', 'description',
    'genres', 'original_publication_year', 'average_rating'
]

print("MOVIE IMPORTANT COLUMN MISSING VALUES")
print(movies[movie_columns].isnull().sum())

print("\nBOOK IMPORTANT COLUMN MISSING VALUES")
print(books[book_columns].isnull().sum())
```
